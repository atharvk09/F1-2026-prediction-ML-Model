```python
from pathlib import Path
import pandas as pd
import fastf1

# Save the dataset in your project's data folder
data_dir = Path("../data")
data_dir.mkdir(parents=True, exist_ok=True)
output_file = data_dir / "race_results_2024.csv"

# Get the 2024 race schedule
schedule_2024 = fastf1.get_event_schedule(2024)
races_2024 = schedule_2024[schedule_2024["Session5"] == "Race"]

# Load any previously saved progress
if output_file.exists():
    collected = pd.read_csv(output_file)
    completed_rounds = set(collected["Round"].unique())
    print(f"Resuming: {len(completed_rounds)} races already saved.")
else:
    collected = pd.DataFrame()
    completed_rounds = set()
    print("Starting a new 2024 dataset.")

# Collect each race and save immediately
for _, event in races_2024.iterrows():
    round_number = int(event["RoundNumber"])

    if round_number in completed_rounds:
        print(f"Skipping round {round_number}: already saved.")
        continue

    event_name = event["EventName"]
    print(f"\nLoading round {round_number}: {event_name}")

    session = fastf1.get_session(2024, round_number, "R")
    session.load()

    race_results = session.results[
        [
            "DriverId",
            "FullName",
            "TeamName",
            "GridPosition",
            "Position",
            "Points",
            "Status",
            "Laps",
        ]
    ].copy()

    race_results["Year"] = 2024
    race_results["Round"] = round_number
    race_results["EventName"] = event_name

    # Append the new race to the CSV immediately
    race_results.to_csv(
        output_file,
        mode="a" if output_file.exists() else "w",
        header=not output_file.exists(),
        index=False,
    )

    completed_rounds.add(round_number)
    print(f"Saved {event_name}: {len(race_results)} driver records.")

print("\nCollection finished.")
print("Saved to:", output_file.resolve())
```

```python
import pandas as pd
from pathlib import Path

data_dir = Path("../data")

# Load each season
results_2022 = pd.read_csv(data_dir / "race_results_2022.csv")
results_2023 = pd.read_csv(data_dir / "race_results_2023.csv")
results_2024 = pd.read_csv(data_dir / "race_results_2024.csv")
results_2025 = pd.read_csv(data_dir / "race_results_2025.csv")

# Combine all seasons
all_results = pd.concat(
    [results_2022, results_2023, results_2024, results_2025],
    ignore_index=True
)

# Sort chronologically
all_results = all_results.sort_values(
    ["Year", "Round", "GridPosition"],
    na_position="last"
).reset_index(drop=True)

# Save the combined dataset
output_file = data_dir / "race_results_2022_2025.csv"
all_results.to_csv(output_file, index=False)

print("Combined dataset saved!")
print("Total driver records:", len(all_results))
print("Years:", sorted(all_results["Year"].unique()))
print("Unique races:", all_results.groupby(["Year", "Round"]).ngroups)

print("\nRecords per season:")
print(all_results.groupby("Year").size())

print("\nFirst five rows:")
print(all_results.head().to_string(index=False))
```

```python
# 1. Check for duplicate driver entries within a race
duplicate_rows = all_results.duplicated(
    subset=["Year", "Round", "DriverId"],
    keep=False
)

print("=== Duplicate Check ===")
print("Duplicate rows:", duplicate_rows.sum())

if duplicate_rows.any():
    print(all_results.loc[duplicate_rows].sort_values(
        ["Year", "Round", "DriverId"]
    ).to_string(index=False))


# 2. Check missing values
print("\n=== Missing Values ===")
print(all_results.isna().sum().to_string())


# 3. Check recorded winners in each race
winner_counts = (
    all_results.assign(IsWinner=all_results["Position"].eq(1))
    .groupby(["Year", "Round", "EventName"])["IsWinner"]
    .sum()
)

print("\n=== Winner Check ===")
print("Races with exactly one recorded winner:",
      (winner_counts == 1).sum(), "/", len(winner_counts))

if (winner_counts != 1).any():
    print("Races needing inspection:")
    print(winner_counts[winner_counts != 1].to_string())


# 4. Inspect finishing-position values
print("\n=== Position Values ===")
print("Unique positions:", sorted(all_results["Position"].dropna().unique()))
print("Missing finishing positions:", all_results["Position"].isna().sum())
```

```python
# Find records with missing grid or finishing positions
missing_positions = all_results[
    all_results["GridPosition"].isna()
    | all_results["Position"].isna()
]

print("Records with missing positions:")
print(
    missing_positions[
        [
            "Year",
            "Round",
            "EventName",
            "DriverId",
            "FullName",
            "TeamName",
            "GridPosition",
            "Position",
            "Status",
            "Laps",
        ]
    ].to_string(index=False)
)
```

```python
# Create a copy so the original combined dataset remains unchanged
model_data = all_results.copy()

# Create the target variable
model_data["Won"] = (model_data["Position"] == 1).astype(int)

# Exclude records without a valid grid position or finishing position
model_data = model_data.dropna(
    subset=["GridPosition", "Position"]
).copy()

print("Total records after filtering:", len(model_data))
print("Winner records:", model_data["Won"].sum())
print("Non-winner records:", (model_data["Won"] == 0).sum())

print("\nWinner count by season:")
print(model_data.groupby("Year")["Won"].sum())

print("\nSample:")
print(
    model_data[
        ["Year", "Round", "FullName", "GridPosition", "Position", "Won"]
    ].head(10).to_string(index=False)
)
```

```python
# Count how many drivers are recorded in each grid position per race
grid_counts = (
    model_data.groupby(["Year", "Round", "GridPosition"])
    .size()
    .rename("DriverCount")
    .reset_index()
)

# Find races where grid positions 1, 2, or 3 occur more than once
unexpected_grid_counts = grid_counts[
    (grid_counts["GridPosition"] <= 3)
    & (grid_counts["DriverCount"] > 1)
]

print("Repeated grid positions among the top three:")
if unexpected_grid_counts.empty:
    print("None found.")
else:
    print(unexpected_grid_counts.to_string(index=False))

# Check the number of drivers assigned to each top-three position
print("\nNumber of records by grid position:")
print(
    model_data["GridPosition"]
    .value_counts()
    .sort_index()
    .head(5)
    .to_string()
)
```

```python
# Pole-sitters: exactly grid position 1
pole_sitters = model_data[model_data["GridPosition"] == 1]

# Top-three starters: positions 1, 2, or 3 only
top_three_starters = model_data[
    model_data["GridPosition"].isin([1, 2, 3])
]

pole_win_rate = pole_sitters["Won"].mean() * 100
top_three_win_rate = top_three_starters["Won"].mean() * 100
overall_win_rate = model_data["Won"].mean() * 100

print("=== Corrected F1 Win-Rate Baselines (2022–2025) ===")

print(
    f"Pole-position win rate: {pole_win_rate:.2f}% "
    f"({int(pole_sitters['Won'].sum())} wins from {len(pole_sitters)} records)"
)

print(
    f"Top-three grid win rate: {top_three_win_rate:.2f}% "
    f"({int(top_three_starters['Won'].sum())} wins from "
    f"{len(top_three_starters)} records)"
)

print(
    f"Overall driver win rate: {overall_win_rate:.2f}% "
    f"({int(model_data['Won'].sum())} wins from {len(model_data)} records)"
)
```

```python
import requests, pandas as pd, time

def to_sec(t):
    if not t: return None
    m, s = t.split(":")
    return int(m) * 60 + float(s)

def get_quali(year):
    rows, offset = [], 0
    while True:
        url = f"https://api.jolpi.ca/ergast/f1/{year}/qualifying.json?limit=100&offset={offset}"
        data = requests.get(url).json()["MRData"]
        races = data["RaceTable"]["Races"]
        for race in races:
            for r in race["QualifyingResults"]:
                rows.append({"Year": year, "Round": int(race["round"]),
                             "DriverId": r["Driver"]["driverId"],
                             "Q1": to_sec(r.get("Q1")), "Q2": to_sec(r.get("Q2")), "Q3": to_sec(r.get("Q3"))})
        offset += 100
        if offset >= int(data["total"]): break
        time.sleep(1)
    return pd.DataFrame(rows)

q2025 = get_quali(2025)
print(q2025.shape)
print(q2025.head())
```

```python
frames = [q2025]
for yr in [2022, 2023, 2024]:
    frames.append(get_quali(yr))
    time.sleep(1)

quali = pd.concat(frames, ignore_index=True)

# fastest valid lap across Q1/Q2/Q3 (NaN if the driver set no time)
quali["BestQualiLap"] = quali[["Q1", "Q2", "Q3"]].min(axis=1)

# percentage gap to the pole lap within each race
pole = quali.groupby(["Year", "Round"])["BestQualiLap"].transform("min")
quali["QualiGapPct"] = (quali["BestQualiLap"] - pole) / pole * 100

quali.to_csv("../data/qualifying_2022_2025.csv", index=False)

print(quali.shape)
print(quali.groupby("Year").size())
print("Missing QualiGapPct:", quali["QualiGapPct"].isna().sum())
print(quali["QualiGapPct"].describe())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
import numpy as np
import pandas as pd

# Model A: grid position only
features_a = ["GridPosition"]

# Model G: grid + qualifying + team + driver form
features_g = [
    "GridPosition",
    "QualiGapPct",
    "QualiMissing",
    "TeamSeasonPointsPerRace",
    "TeamPrevious5RacePoints",
    "PreviousWinRate",
    "Previous5RaceWins"
]

def evaluate_model(train_data, test_data, features, model_name):
    # Prepare data
    X_train = train_data[features]
    y_train = train_data["Won"]
    X_test = test_data[features]

    # Build and train model
    model = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(max_iter=2000))
    ])

    model.fit(X_train, y_train)

    # Predict and normalize probabilities within each race
    predictions = test_data[["Year", "Round", "EventName", "FullName",
                             "GridPosition", "Won"]].copy()

    predictions["Probability"] = model.predict_proba(X_test)[:, 1]
    predictions["RaceWinProbability"] = (
        predictions["Probability"]
        / predictions.groupby(["Year", "Round"])["Probability"].transform("sum")
    )

    # Choose one predicted winner per race
    winners = predictions.loc[
        predictions.groupby(["Year", "Round"])["RaceWinProbability"].idxmax()
    ]

    accuracy = winners["Won"].mean()

    # Race-level log loss and multiclass Brier score
    log_losses = []
    brier_scores = []

    for _, race in predictions.groupby(["Year", "Round"]):
        probabilities = race["RaceWinProbability"].to_numpy()
        actual = race["Won"].to_numpy()

        winner_probability = probabilities[actual == 1][0]
        log_losses.append(-np.log(np.clip(winner_probability, 1e-15, 1.0)))
        brier_scores.append(np.sum((probabilities - actual) ** 2))

    return {
        "Model": model_name,
        "Races": len(winners),
        "Correct": int(winners["Won"].sum()),
        "Accuracy": accuracy,
        "LogLoss": np.mean(log_losses),
        "Brier": np.mean(brier_scores)
    }

results = []

# Chronological test splits
splits = [
    (2022, [2023]),
    (2022, [2023, 2024]),
    (2022, [2023, 2024, 2025])
]

# Evaluate each season separately with only earlier seasons for training
for test_year in [2023, 2024, 2025]:
    train_data = evaluation_data_fair[
        evaluation_data_fair["Year"] < test_year
    ].copy()

    test_data = evaluation_data_fair[
        evaluation_data_fair["Year"] == test_year
    ].copy()

    for model_name, features in [
        ("Model A", features_a),
        ("Model G", features_g)
    ]:
        result = evaluate_model(
            train_data,
            test_data,
            features,
            model_name
        )
        result["TestYear"] = test_year
        results.append(result)

results_df = pd.DataFrame(results)

print(
    results_df[
        ["TestYear", "Model", "Races", "Correct", "Accuracy", "LogLoss", "Brier"]
    ].to_string(index=False, formatters={
        "Accuracy": "{:.2%}".format,
        "LogLoss": "{:.4f}".format,
        "Brier": "{:.4f}".format
    })
)
```

```python
import numpy as np

decay_data = model_data.sort_values(
    ["DriverId", "Year", "Round"]
).copy()

half_life = 5

def decayed_win_rate(wins):
    values = wins.to_numpy(dtype=float)
    result = np.zeros(len(values))

    for i in range(len(values)):
        past_wins = values[:i][::-1]
        if len(past_wins) == 0:
            result[i] = 0.0
        else:
            weights = 0.5 ** (np.arange(len(past_wins)) / half_life)
            result[i] = np.sum(past_wins * weights) / np.sum(weights)

    return result

decay_data["DecayedWinRate"] = (
    decay_data.groupby("DriverId")["Won"]
    .transform(decayed_win_rate)
)
```

```python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression


feature_sets = {
    "Grid only": [
        "GridPosition"
    ],
    "Grid + qualifying": [
        "GridPosition",
        "QualiGapPct",
        "QualiMissing"
    ],
    "Grid + qualifying + team": [
        "GridPosition",
        "QualiGapPct",
        "QualiMissing",
        "TeamSeasonPointsPerRace",
        "TeamPrevious5RacePoints"
    ]
}


def evaluate_race_model(features, test_year):
    train = evaluation_data_fair[
        evaluation_data_fair["Year"] < test_year
    ].copy()

    test = evaluation_data_fair[
        evaluation_data_fair["Year"] == test_year
    ].copy()

    model = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(max_iter=2000))
    ])

    model.fit(train[features], train["Won"])

    test["RawProbability"] = model.predict_proba(
        test[features]
    )[:, 1]

    race_log_losses = []
    correct = 0
    total = 0

    for _, race in test.groupby(["Year", "Round"]):
        probabilities = (
            race["RawProbability"] / race["RawProbability"].sum()
        )

        winner_indices = race.index[race["Won"] == 1]

        if len(winner_indices) != 1:
            raise ValueError(
                f"Expected one winner, found {len(winner_indices)}."
            )

        winner_probability = probabilities.loc[winner_indices[0]]
        race_log_losses.append(
            -np.log(max(float(winner_probability), 1e-15))
        )

        predicted_index = probabilities.idxmax()
        correct += int(race.loc[predicted_index, "Won"] == 1)
        total += 1

    return {
        "LogLoss": np.mean(race_log_losses),
        "Accuracy": correct / total,
        "Correct": correct,
        "Races": total
    }


results = []

for model_name, features in feature_sets.items():
    for year in [2023, 2024, 2025]:
        metrics = evaluate_race_model(features, year)

        results.append({
            "Model": model_name,
            "TestYear": year,
            "Races": metrics["Races"],
            "Correct": metrics["Correct"],
            "Accuracy": metrics["Accuracy"],
            "LogLoss": metrics["LogLoss"]
        })

results_df = pd.DataFrame(results)

display(
    results_df.round({
        "Accuracy": 4,
        "LogLoss": 4
    })
)

print("\nAverage metrics by model:")
display(
    results_df.groupby("Model")[
        ["Accuracy", "LogLoss"]
    ].mean().sort_values("LogLoss").round(4)
)
```

```python
import numpy as np
import pandas as pd

from scipy.optimize import minimize
from scipy.special import logsumexp, softmax
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Use the same five features as our current best model.
features = [
    "GridPosition",
    "QualiGapPct",
    "QualiMissing",
    "TeamSeasonPointsPerRace",
    "CurrentSeasonPrevious3RaceAveragePosition"
]

def prepare_race_groups(data, transformed_features):
    """Create one feature matrix and winner index per race."""
    groups = []

    for _, race in data.groupby(["Year", "Round"], sort=True):
        X_race = transformed_features.loc[race.index].to_numpy()
        winner_indices = np.flatnonzero(race["Won"].to_numpy() == 1)

        if len(winner_indices) != 1:
            raise ValueError(
                f"Expected exactly one winner in "
                f"{race['Year'].iloc[0]} Round {race['Round'].iloc[0]}"
            )

        groups.append((X_race, winner_indices[0]))

    return groups


def conditional_loss_and_gradient(beta, groups, regularization=0.01):
    """Race-level negative log likelihood and its gradient."""
    total_loss = 0.0
    total_gradient = np.zeros_like(beta)

    for X_race, winner_index in groups:
        scores = X_race @ beta
        probabilities = softmax(scores)

        total_loss += (
            logsumexp(scores) - scores[winner_index]
        )

        total_gradient += (
            probabilities @ X_race - X_race[winner_index]
        )

    # Small L2 penalty to discourage excessively large coefficients.
    total_loss += 0.5 * regularization * np.dot(beta, beta)
    total_gradient += regularization * beta

    return total_loss, total_gradient


def fit_conditional_model(train_data, feature_names):
    """Fit preprocessing and race-level conditional logistic regression."""
    imputer = SimpleImputer(strategy="median")
    scaler = StandardScaler()

    X_train_imputed = imputer.fit_transform(train_data[feature_names])
    X_train_scaled = scaler.fit_transform(X_train_imputed)

    # Preserve row alignment for race grouping.
    transformed = pd.DataFrame(
        X_train_scaled,
        index=train_data.index,
        columns=feature_names
    )

    groups = prepare_race_groups(train_data, transformed)

    result = minimize(
        fun=lambda beta: conditional_loss_and_gradient(beta, groups),
        x0=np.zeros(len(feature_names)),
        jac=True,
        method="L-BFGS-B",
        options={"maxiter": 2000, "ftol": 1e-10}
    )

    if not result.success:
        print("Optimizer warning:", result.message)

    return imputer, scaler, result.x


def predict_race_probabilities(data, feature_names, imputer, scaler, beta):
    """Calculate probabilities that sum to one within each race."""
    X_imputed = imputer.transform(data[feature_names])
    X_scaled = scaler.transform(X_imputed)

    predictions = data[["Year", "Round", "Won"]].copy()
    predictions["WinProbability"] = np.nan

    predictions["RawScore"] = X_scaled @ beta

    for _, race in predictions.groupby(["Year", "Round"], sort=False):
        probabilities = softmax(race["RawScore"].to_numpy())
        predictions.loc[race.index, "WinProbability"] = probabilities

    return predictions


results = []
conditional_predictions = []

for test_year in [2023, 2024, 2025]:
    train = evaluation_current[
        evaluation_current["Year"] < test_year
    ].copy()

    test = evaluation_current[
        evaluation_current["Year"] == test_year
    ].copy()

    imputer, scaler, beta = fit_conditional_model(train, features)

    predictions = predict_race_probabilities(
        test, features, imputer, scaler, beta
    )

    predictions["FullName"] = test["FullName"]
    predictions["EventName"] = test["EventName"]

    predictions["PredictedWinner"] = False

    for _, race in predictions.groupby(["Year", "Round"], sort=False):
        best_index = race["WinProbability"].idxmax()
        predictions.loc[best_index, "PredictedWinner"] = True

    race_log_losses = []

    for _, race in predictions.groupby(["Year", "Round"], sort=True):
        winner_probability = race.loc[
            race["Won"] == 1, "WinProbability"
        ].iloc[0]

        race_log_losses.append(
            -np.log(max(winner_probability, 1e-15))
        )

    accuracy = (
        predictions["PredictedWinner"] & predictions["Won"].eq(1)
    ).sum() / predictions.groupby(["Year", "Round"]).ngroups

    results.append({
        "TestYear": test_year,
        "Races": predictions.groupby(["Year", "Round"]).ngroups,
        "Correct": int(
            (
                predictions["PredictedWinner"]
                & predictions["Won"].eq(1)
            ).sum()
        ),
        "Accuracy": accuracy,
        "LogLoss": np.mean(race_log_losses)
    })

    conditional_predictions.append(predictions)

conditional_results = pd.DataFrame(results)

display(
    conditional_results.assign(
        Accuracy=lambda df: (df["Accuracy"] * 100).round(2)
    ).round({"LogLoss": 4})
)

print("\nAverage across test seasons:")
print(
    "Accuracy:",
    round(conditional_results["Accuracy"].mean() * 100, 2),
    "%"
)
print(
    "Log loss:",
    round(conditional_results["LogLoss"].mean(), 4)
)
```

```python
import requests
import pandas as pd
from pathlib import Path

base_url = "https://api.jolpi.ca/ergast/f1/2026/qualifying.json"

all_races = []
offset = 0
page_size = 100

while True:
    response = requests.get(
        base_url,
        params={"limit": page_size, "offset": offset},
        timeout=30
    )
    response.raise_for_status()

    mr_data = response.json()["MRData"]
    races_page = mr_data["RaceTable"]["Races"]

    all_races.extend(races_page)

    total_records = int(mr_data["total"])
    returned_limit = int(mr_data["limit"])

    print(
        f"Fetched {len(races_page)} races, "
        f"{sum(len(r.get('QualifyingResults', [])) for r in races_page)} "
        f"qualifying records (offset {offset})"
    )

    if not races_page:
        break

    offset += returned_limit

    if offset >= total_records:
        break


def lap_time_to_seconds(value):
    """Convert an F1 lap time such as '1:30.123' to seconds."""
    if not value or pd.isna(value):
        return None

    parts = str(value).split(":")

    try:
        if len(parts) == 2:
            return int(parts[0]) * 60 + float(parts[1])
        return float(parts[0])
    except (ValueError, TypeError):
        return None


rows = []

for race in all_races:
    for result in race.get("QualifyingResults", []):
        driver = result["Driver"]
        constructor = result["Constructor"]

        q1 = result.get("Q1")
        q2 = result.get("Q2")
        q3 = result.get("Q3")

        available_times = [
            lap_time_to_seconds(t)
            for t in [q1, q2, q3]
        ]
        available_times = [t for t in available_times if t is not None]

        best_time = min(available_times) if available_times else None

        rows.append({
            "DriverId": driver["driverId"],
            "FullName": (
                f'{driver["givenName"]} {driver["familyName"]}'
            ),
            "TeamName": constructor["name"],
            "Year": 2026,
            "Round": int(race["round"]),
            "EventName": race["raceName"],
            "QualifyingPosition": pd.to_numeric(
                result.get("position"), errors="coerce"
            ),
            "Q1": q1,
            "Q2": q2,
            "Q3": q3,
            "BestQualiTimeSeconds": best_time
        })


qualifying_2026 = pd.DataFrame(rows)

# Calculate the percentage gap from the fastest available qualifying time
qualifying_2026["QualiGapPct"] = (
    qualifying_2026.groupby(["Year", "Round"])["BestQualiTimeSeconds"]
    .transform(
        lambda times: (
            times / times.min() - 1
        ) * 100
    )
)

qualifying_2026["QualiMissing"] = (
    qualifying_2026["QualiGapPct"].isna().astype(int)
)

output_path = Path("../data/qualifying_2026.csv")
qualifying_2026.to_csv(output_path, index=False)

print("\nSaved:", output_path)
print("Qualifying records:", len(qualifying_2026))
print("Unique rounds:", qualifying_2026["Round"].nunique())

print("\nSingapore qualifying data:")
print(
    qualifying_2026.loc[
        qualifying_2026["EventName"].str.contains(
            "Singapore", case=False, na=False
        ),
        [
            "FullName",
            "TeamName",
            "QualifyingPosition",
            "Q1",
            "Q2",
            "Q3",
            "QualiGapPct",
            "QualiMissing"
        ]
    ]
    .sort_values("QualifyingPosition")
    .to_string(index=False)
)
```

```python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# --------------------------------------------------
# 1. Define the five features from our best model
# --------------------------------------------------

features = [
    "GridPosition",
    "QualiGapPct",
    "QualiMissing",
    "TeamSeasonPointsPerRace",
    "CurrentSeasonPrevious3RaceAveragePosition"
]

# --------------------------------------------------
# 2. Prepare the historical training data (2022-2025)
# --------------------------------------------------

current_form = feature_data[
    [
        "Year",
        "Round",
        "DriverId",
        "CurrentSeasonPrevious3RaceAveragePosition"
    ]
].copy()

training_data = model_data.drop(
    columns=["CurrentSeasonPrevious3RaceAveragePosition"],
    errors="ignore"
).merge(
    current_form,
    on=["Year", "Round", "DriverId"],
    how="left",
    validate="one_to_one"
)

# Match the valid-grid approach used in our earlier analysis
training_data = training_data[
    training_data["GridPosition"].between(1, 20)
].copy()

X_train = training_data[features]
y_train = training_data["Won"].astype(int)

print("Training rows:", len(training_data))
print("Training seasons:", sorted(training_data["Year"].unique()))
print("Training winners:", int(y_train.sum()))

# --------------------------------------------------
# 3. Calculate 2026 driver form using Rounds 1-16 only
# --------------------------------------------------

previous_2026 = race_results_2026[
    race_results_2026["Round"] < 17
].copy()

previous_2026 = previous_2026.sort_values(
    ["DriverId", "Round"]
)

def previous_three_average(series):
    return series.shift(1).rolling(
        window=3,
        min_periods=1
    ).mean()

previous_2026[
    "CurrentSeasonPrevious3RaceAveragePosition"
] = (
    previous_2026.groupby("DriverId")["Position"]
    .transform(previous_three_average)
)

# Keep the most recent historical value for each driver
driver_form_2026 = (
    previous_2026.sort_values(["DriverId", "Round"])
    .groupby("DriverId", as_index=False)
    .tail(1)[
        ["DriverId", "CurrentSeasonPrevious3RaceAveragePosition"]
    ]
)

# --------------------------------------------------
# 4. Calculate team form using Rounds 1-16 only
# --------------------------------------------------

team_race_points = (
    previous_2026.groupby(
        ["Year", "Round", "TeamName"],
        as_index=False
    )["Points"]
    .sum()
    .rename(columns={"Points": "TeamRacePoints"})
    .sort_values(["TeamName", "Round"])
)

team_race_points["TeamSeasonPointsPerRace"] = (
    team_race_points.groupby("TeamName")["TeamRacePoints"]
    .transform(
        lambda s: s.shift(1).expanding(min_periods=1).mean()
    )
)

# Each team's value after Round 16 is its average points
# per previous race, excluding the Singapore Grand Prix.
team_form_2026 = (
    team_race_points.groupby("TeamName", as_index=False)
    .tail(1)[["TeamName", "TeamSeasonPointsPerRace"]]
)

# --------------------------------------------------
# 5. Prepare Singapore's pre-race prediction rows
# --------------------------------------------------

singapore = race_results_2026[
    race_results_2026["Round"] == 17
][
    ["DriverId", "FullName", "TeamName", "GridPosition"]
].copy()

singapore = singapore.merge(
    qualifying_2026.loc[
        qualifying_2026["Round"] == 17,
        ["DriverId", "QualiGapPct", "QualiMissing"]
    ],
    on="DriverId",
    how="left",
    validate="one_to_one"
)

singapore = singapore.merge(
    driver_form_2026,
    on="DriverId",
    how="left",
    validate="one_to_one"
)

singapore = singapore.merge(
    team_form_2026,
    on="TeamName",
    how="left",
    validate="many_to_one"
)

# Check that we have all 22 drivers and the expected features
assert len(singapore) == 22, (
    f"Expected 22 Singapore drivers, found {len(singapore)}"
)

assert singapore["DriverId"].is_unique, (
    "Duplicate driver IDs in Singapore prediction data"
)

assert singapore["GridPosition"].notna().all(), (
    "At least one Singapore driver has no grid position"
)

print("\nSingapore prediction data checks:")
print("Drivers:", len(singapore))
print("Unique drivers:", singapore["DriverId"].nunique())
print("Missing qualifying gaps:", int(singapore["QualiGapPct"].isna().sum()))
print(
    "Missing driver-form values:",
    int(singapore["CurrentSeasonPrevious3RaceAveragePosition"].isna().sum())
)
print(
    "Missing team-form values:",
    int(singapore["TeamSeasonPointsPerRace"].isna().sum())
)

# --------------------------------------------------
# 6. Train on 2022-2025 and predict Singapore
# --------------------------------------------------

model_2026 = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=2000))
])

model_2026.fit(X_train[features], y_train)

# Predict each driver's probability, then normalize within the race
raw_probabilities = model_2026.predict_proba(
    singapore[features]
)[:, 1]

if raw_probabilities.sum() == 0:
    raise ValueError("All predicted probabilities are zero.")

singapore["PredictedWinProbability"] = (
    raw_probabilities / raw_probabilities.sum()
)

singapore = singapore.sort_values(
    "PredictedWinProbability",
    ascending=False
).reset_index(drop=True)

singapore["PredictedWinProbabilityPct"] = (
    singapore["PredictedWinProbability"] * 100
)

print("\n=== 2026 SINGAPORE GP: MODEL PREDICTION ===")
print(
    singapore[
        [
            "FullName",
            "TeamName",
            "GridPosition",
            "QualiGapPct",
            "TeamSeasonPointsPerRace",
            "CurrentSeasonPrevious3RaceAveragePosition",
            "PredictedWinProbabilityPct"
        ]
    ].to_string(
        index=False,
        formatters={
            "QualiGapPct": lambda x: (
                f"{x:.3f}" if pd.notna(x) else "Missing"
            ),
            "TeamSeasonPointsPerRace": lambda x: (
                f"{x:.2f}" if pd.notna(x) else "Missing"
            ),
            "CurrentSeasonPrevious3RaceAveragePosition": lambda x: (
                f"{x:.2f}" if pd.notna(x) else "Missing"
            ),
            "PredictedWinProbabilityPct": lambda x: f"{x:.2f}%"
        }
    )
)

print("\nPredicted winner:", singapore.iloc[0]["FullName"])
print(
    "Predicted win probability:",
    f'{singapore.iloc[0]["PredictedWinProbabilityPct"]:.2f}%'
)
```

```python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# --------------------------------------------------
# 1. Train the existing model on 2022-2025 only
# --------------------------------------------------

features = [
    "GridPosition",
    "QualiGapPct",
    "QualiMissing",
    "TeamSeasonPointsPerRace",
    "CurrentSeasonPrevious3RaceAveragePosition"
]

historical_form = feature_data[
    [
        "Year",
        "Round",
        "DriverId",
        "CurrentSeasonPrevious3RaceAveragePosition"
    ]
].copy()

training_data = model_data.drop(
    columns=["CurrentSeasonPrevious3RaceAveragePosition"],
    errors="ignore"
).merge(
    historical_form,
    on=["Year", "Round", "DriverId"],
    how="left",
    validate="one_to_one"
)

training_data = training_data[
    training_data["GridPosition"].between(1, 20)
].copy()

model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=2000))
])

model.fit(training_data[features], training_data["Won"].astype(int))

print("Training years:", sorted(training_data["Year"].unique()))
print("Training rows:", len(training_data))


# --------------------------------------------------
# 2. Prepare the 2026 data
# --------------------------------------------------

results_2026 = race_results_2026.copy()
results_2026["Position"] = pd.to_numeric(
    results_2026["Position"], errors="coerce"
)

# Use each driver's previous completed race results.
# shift(1) ensures the current race's result is excluded.
results_2026 = results_2026.sort_values(
    ["DriverId", "Round"]
).copy()

results_2026["Previous3RaceAveragePosition"] = (
    results_2026.groupby("DriverId")["Position"]
    .transform(lambda s: s.shift(1).rolling(3, min_periods=1).mean())
)

# Calculate team points per race using only earlier races.
team_race_points = (
    results_2026.groupby(["TeamName", "Round"], as_index=False)["Points"]
    .sum()
    .rename(columns={"Points": "TeamRacePoints"})
    .sort_values(["TeamName", "Round"])
)

team_race_points["TeamSeasonPointsPerRace"] = (
    team_race_points.groupby("TeamName")["TeamRacePoints"]
    .transform(lambda s: s.shift(1).expanding().mean())
)

# Qualifying gaps for the 2026 races
quali_2026 = qualifying_2026.copy()

# Calculate the fastest available qualifying time per driver.
time_columns = ["Q1", "Q2", "Q3"]
for col in time_columns:
    if col in quali_2026.columns:
        quali_2026[col] = pd.to_timedelta(
            quali_2026[col], errors="coerce"
        ).dt.total_seconds()

available_time_columns = [
    col for col in time_columns if col in quali_2026.columns
]

quali_2026["BestQualifyingTime"] = quali_2026[
    available_time_columns
].min(axis=1)

quali_2026["FastestRaceQualifyingTime"] = (
    quali_2026.groupby("Round")["BestQualifyingTime"].transform("min")
)

quali_2026["QualiGapPct"] = (
    (
        quali_2026["BestQualifyingTime"]
        - quali_2026["FastestRaceQualifyingTime"]
    )
    / quali_2026["FastestRaceQualifyingTime"]
    * 100
)

quali_2026["QualiMissing"] = (
    quali_2026["QualiGapPct"].isna().astype(int)
)

# Check which qualifying identifiers are available
print("\nQualifying columns:", quali_2026.columns.tolist())
```

```python
# Prepare race-by-race predictions using only pre-race information

predictions_2026 = []

for round_num in range(1, 17):

    # Current race entries, including grid position
    race = results_2026[
        results_2026["Round"] == round_num
    ].copy()

    # Driver form available before this race
    driver_form = race[
        ["DriverId", "Previous3RaceAveragePosition"]
    ].copy()

    # Team form available before this race
    team_form = team_race_points[
        team_race_points["Round"] == round_num
    ][["TeamName", "TeamSeasonPointsPerRace"]].copy()

    # Qualifying information for this race
    race_quali = quali_2026[
        quali_2026["Round"] == round_num
    ][["DriverId", "QualiGapPct", "QualiMissing"]].copy()

    # Combine the pre-race features
    race_features = race[
        ["Year", "Round", "DriverId", "FullName", "TeamName", "GridPosition"]
    ].merge(
        driver_form,
        on="DriverId",
        how="left",
        validate="one_to_one"
    ).merge(
        team_form,
        on="TeamName",
        how="left",
        validate="many_to_one"
    ).merge(
        race_quali,
        on="DriverId",
        how="left",
        validate="one_to_one"
    )

    # Rename the driver-form column to match the trained model
    race_features = race_features.rename(columns={
        "Previous3RaceAveragePosition":
            "CurrentSeasonPrevious3RaceAveragePosition"
    })

    # Predict win probabilities
    raw_probabilities = model.predict_proba(
        race_features[features]
    )[:, 1]

    # Normalize probabilities so each race sums to 100%
    race_features["PredictedWinProbabilityPct"] = (
        raw_probabilities / raw_probabilities.sum() * 100
    )

    # Actual results are used only for evaluation, never as predictors
    actual = race[
        ["DriverId", "Position"]
    ].copy()

    actual["ActualPosition"] = pd.to_numeric(
        actual["Position"], errors="coerce"
    )

    race_features = race_features.merge(
        actual[["DriverId", "ActualPosition"]],
        on="DriverId",
        how="left",
        validate="one_to_one"
    )

    race_features["PredictedRank"] = (
        race_features["PredictedWinProbabilityPct"]
        .rank(method="min", ascending=False)
        .astype(int)
    )

    predictions_2026.append(race_features)

# Combine all 16 races
predictions_2026 = pd.concat(
    predictions_2026,
    ignore_index=True
)

print("Predicted races:", predictions_2026["Round"].nunique())
print("Driver-race predictions:", len(predictions_2026))
print(
    "Probability sum by race (should be approximately 100):"
)
display(
    predictions_2026.groupby("Round")[
        "PredictedWinProbabilityPct"
    ].sum().round(2)
)
```

```python
import numpy as np
import pandas as pd

from lightgbm import LGBMClassifier
from sklearn.base import clone
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Keep the same five features used by our existing models.
comparison_features = [
    "GridPosition",
    "QualiGapPct",
    "QualiMissing",
    "TeamSeasonPointsPerRace",
    "CurrentSeasonPrevious3RaceAveragePosition"
]

# Separate LightGBM model for the 2026 experiment.
lightgbm_2026_template = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("classifier", LGBMClassifier(
        objective="binary",
        n_estimators=200,
        learning_rate=0.03,
        num_leaves=7,
        max_depth=3,
        min_child_samples=20,
        reg_lambda=1.0,
        random_state=42,
        verbosity=-1
    ))
])

# Give more weight to recent historical seasons, as in Model B.
season_weights = {
    2022: 0.5,
    2023: 0.75,
    2024: 1.0,
    2025: 1.5
}

lightgbm_2026_results = []

# Predict each round using only information available before that round.
for current_round in sorted(predictions_2026["Round"].unique()):

    # Historical data from 2022-2025.
    historical_fold = training_data.copy()
    historical_fold["Target"] = historical_fold["Won"].astype(int)
    historical_fold["SampleWeight"] = (
        historical_fold["Year"].map(season_weights).fillna(1.0)
    )

    # Only include 2026 races completed before the current round.
    previous_2026 = predictions_2026[
        predictions_2026["Round"] < current_round
    ].copy()

    if not previous_2026.empty:
        previous_2026["Target"] = (
            previous_2026["ActualPosition"] == 1
        ).astype(int)
        previous_2026["SampleWeight"] = 3.0

        previous_training = previous_2026[
            comparison_features + ["Target", "SampleWeight"]
        ].copy()

        historical_training = historical_fold[
            comparison_features + ["Target", "SampleWeight"]
        ].copy()

        combined_training = pd.concat(
            [historical_training, previous_training],
            ignore_index=True
        )
    else:
        combined_training = historical_fold[
            comparison_features + ["Target", "SampleWeight"]
        ].copy()

    current_race = predictions_2026[
        predictions_2026["Round"] == current_round
    ].copy()

    fitted_lightgbm = clone(lightgbm_2026_template)

    fitted_lightgbm.fit(
        combined_training[comparison_features],
        combined_training["Target"],
        classifier__sample_weight=combined_training["SampleWeight"]
    )

    raw_probabilities = fitted_lightgbm.predict_proba(
        current_race[comparison_features]
    )[:, 1]

    current_race["LightGBMRawProbability"] = raw_probabilities

    # Normalize scores so each race's driver probabilities sum to 100%.
    probability_total = raw_probabilities.sum()

    if probability_total > 0:
        current_race["LightGBMWinProbabilityPct"] = (
            raw_probabilities / probability_total * 100
        )
    else:
        current_race["LightGBMWinProbabilityPct"] = (
            100 / len(current_race)
        )

    current_race["LightGBMPredictedRank"] = (
        current_race["LightGBMWinProbabilityPct"]
        .rank(method="min", ascending=False)
    )

    lightgbm_2026_results.append(current_race)

# Create a new results dataframe without changing existing predictions.
lightgbm_2026_predictions = pd.concat(
    lightgbm_2026_results,
    ignore_index=True
)

print("LightGBM 2026 predictions created.")
print("Rows:", len(lightgbm_2026_predictions))
print("Rounds:", lightgbm_2026_predictions["Round"].nunique())
print("Drivers per round:")
display(
    lightgbm_2026_predictions.groupby("Round").size().to_frame("Drivers")
)

print("\nFirst-round prediction preview:")
display(
    lightgbm_2026_predictions[
        [
            "Round",
            "FullName",
            "ActualPosition",
            "LightGBMWinProbabilityPct",
            "LightGBMPredictedRank"
        ]
    ]
    .query("Round == 1")
    .sort_values("LightGBMPredictedRank")
    .head(10)
)
```

```python
import numpy as np
import pandas as pd

# Build a comparison table using the same driver and round identifiers.
model_a_eval = predictions_2026[
    ["Year", "Round", "DriverId", "ActualPosition",
     "PredictedWinProbabilityPct", "PredictedRank"]
].copy()

model_b_eval = adaptive_predictions[
    ["Year", "Round", "DriverId", "AdaptiveWinProbabilityPct",
     "AdaptivePredictedRank"]
].copy()

model_c_eval = lightgbm_2026_predictions[
    ["Year", "Round", "DriverId", "LightGBMWinProbabilityPct",
     "LightGBMPredictedRank"]
].copy()

comparison_2026 = model_a_eval.merge(
    model_b_eval,
    on=["Year", "Round", "DriverId"],
    how="inner",
    validate="one_to_one"
).merge(
    model_c_eval,
    on=["Year", "Round", "DriverId"],
    how="inner",
    validate="one_to_one"
)

# Confirm that all three models cover the same drivers and rounds.
print("Rows in comparison:", len(comparison_2026))
print("Rounds:", comparison_2026["Round"].nunique())
print("Drivers per round:")
display(comparison_2026.groupby("Round").size().to_frame("Drivers"))

# Evaluate each model at the race level.
model_probability_columns = {
    "Model A: Logistic Regression": "PredictedWinProbabilityPct",
    "Model B: Adaptive Logistic": "AdaptiveWinProbabilityPct",
    "Model C: LightGBM": "LightGBMWinProbabilityPct"
}

model_rank_columns = {
    "Model A: Logistic Regression": "PredictedRank",
    "Model B: Adaptive Logistic": "AdaptivePredictedRank",
    "Model C: LightGBM": "LightGBMPredictedRank"
}

evaluation_rows = []

for model_name in model_probability_columns:
    probability_col = model_probability_columns[model_name]
    rank_col = model_rank_columns[model_name]

    winners = comparison_2026[
        comparison_2026["ActualPosition"] == 1
    ].copy()

    # Convert percentage probabilities to proportions for log loss.
    winner_probabilities = (
        winners[probability_col].to_numpy(dtype=float) / 100
    )

    winner_probabilities = np.clip(
        winner_probabilities, 1e-15, 1.0
    )

    correct = int((winners[rank_col] == 1).sum())
    race_count = winners[["Year", "Round"]].drop_duplicates().shape[0]

    evaluation_rows.append({
        "Model": model_name,
        "Races": race_count,
        "CorrectWinners": correct,
        "WinnerAccuracyPct": round(correct / race_count * 100, 2),
        "RaceLevelLogLoss": round(
            -np.log(winner_probabilities).mean(), 4
        )
    })

comparison_2026_summary = pd.DataFrame(evaluation_rows)

print("\n2026 model comparison")
display(comparison_2026_summary)

print("\nCorrect winner predictions by round")
winner_rows = comparison_2026[
    comparison_2026["ActualPosition"] == 1
].copy()

winner_rows["Model A Correct"] = winner_rows["PredictedRank"] == 1
winner_rows["Model B Correct"] = (
    winner_rows["AdaptivePredictedRank"] == 1
)
winner_rows["Model C Correct"] = (
    winner_rows["LightGBMPredictedRank"] == 1
)

display(
    winner_rows[
        [
            "Round",
            "ActualPosition",
            "Model A Correct",
            "Model B Correct",
            "Model C Correct"
        ]
    ].sort_values("Round")
)
```

```python
import numpy as np
import pandas as pd

# Check that every model's probabilities sum to approximately 100%
# within each race.
probability_checks = []

for model_name, probability_col in {
    "Model A": "PredictedWinProbabilityPct",
    "Model B": "AdaptiveWinProbabilityPct",
    "Model C": "LightGBMWinProbabilityPct"
}.items():

    race_totals = comparison_2026.groupby(
        ["Year", "Round"]
    )[probability_col].sum()

    probability_checks.append({
        "Model": model_name,
        "MinimumRaceTotalPct": round(race_totals.min(), 4),
        "MaximumRaceTotalPct": round(race_totals.max(), 4),
        "AllRacesSumTo100Pct": bool(
            np.allclose(race_totals.to_numpy(), 100, atol=0.02)
        )
    })

print("Probability validation")
display(pd.DataFrame(probability_checks))

# Evaluate top-three overlap for each race.
podium_results = []

for model_name, rank_col in {
    "Model A": "PredictedRank",
    "Model B": "AdaptivePredictedRank",
    "Model C": "LightGBMPredictedRank"
}.items():

    race_scores = []

    for (year, round_number), race in comparison_2026.groupby(
        ["Year", "Round"]
    ):

        predicted_top_three = set(
            race.loc[race[rank_col] <= 3, "DriverId"]
        )

        actual_podium = set(
            race.loc[race["ActualPosition"].isin([1, 2, 3]), "DriverId"]
        )

        race_scores.append(
            len(predicted_top_three & actual_podium)
        )

    podium_results.append({
        "Model": model_name,
        "AverageActualPodiumDriversInPredictedTop3": round(
            np.mean(race_scores), 3
        ),
        "TotalPodiumHitsAcrossRaces": int(sum(race_scores)),
        "MaximumPossibleHits": len(race_scores) * 3
    })

print("\nPredicted top-three versus actual podium")
display(pd.DataFrame(podium_results))
```
